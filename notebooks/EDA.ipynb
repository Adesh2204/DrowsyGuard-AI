# DrowsyGuard AI - EDA

This notebook explores eye-state image distribution and temporal sequence characteristics.

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

ROOT = Path('..').resolve()
eye_root = ROOT / 'data' / 'eye_state'
seq_root = ROOT / 'data' / 'sequences'
eye_root, seq_root
```

```python
def count_images(path):
    exts = {'.jpg', '.jpeg', '.png', '.bmp', '.webp'}
    return sum(1 for p in path.rglob('*') if p.suffix.lower() in exts)

for cls in ['open', 'closed', 'half']:
    cls_path = eye_root / cls
    print(cls, count_images(cls_path) if cls_path.exists() else 0)
```

```python
files = sorted(seq_root.glob('*.npz'))
print('Temporal sequence files:', len(files))
if files:
    sample = np.load(files[0])
    key = 'sequence' if 'sequence' in sample else 'x'
    print('Sample shape:', sample[key].shape)
```
